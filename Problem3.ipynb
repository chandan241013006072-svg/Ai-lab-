```python
class CityNetwork:
    def __init__(self, cities):
        # Initialize graph representation using an adjacency list
        self.graph = {city: [] for city in cities}

    def add_road(self, city1, city2):
        """Adds an undirected road (edge) between two cities."""
        if city1 in self.graph and city2 in self.graph:
            self.graph[city1].append(city2)
            self.graph[city2].append(city1)

    def route_exists(self, start_city, target_city):
        """Checks whether a route exists between start_city and target_city using DFS."""
        if start_city not in self.graph or target_city not in self.graph:
            return False

        visited = set()

        def dfs(current_city):
            if current_city == target_city:
                return True
            visited.add(current_city)
            for neighbor in self.graph[current_city]:
                if neighbor not in visited:
                    if dfs(neighbor):
                        return True
            return False

        return dfs(start_city)

    def find_connected_components(self):
        """Finds all connected components (separate groups of connected cities)."""
        visited = set()
        components = []

        def dfs_component(city, current_component):
            visited.add(city)
            current_component.append(city)
            for neighbor in self.graph[city]:
                if neighbor not in visited:
                    dfs_component(neighbor, current_component)

        for city in self.graph:
            if city not in visited:
                component = []
                dfs_component(city, component)
                components.append(component)

        return components


# Example Usage
if __name__ == "__main__":
    cities = ["A", "B", "C", "D", "E", "F", "G"]
    network = CityNetwork(cities)

    # Group 1: A - B - C
    network.add_road("A", "B")
    network.add_road("B", "C")

    # Group 2: D - E
    network.add_road("D", "E")

    # Group 3: F (isolated), G (isolated)

    # 1. Check if routes exist
    print("Route between A and C:", network.route_exists("A", "C"))  # True
    print("Route between A and D:", network.route_exists("A", "D"))  # False

    # 2. Find connected components
    components = network.find_connected_components()
    print("\nConnected Components:")
    for idx, comp in enumerate(components, 1):
        print(f"Group {idx}: {comp}")
```
